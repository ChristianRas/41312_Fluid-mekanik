```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import linregress
```


![alt text](table1.png)


```python
time_measurements = np.array([11.6, 8.3, 6.3, 4.5, 4.0, 3.4])  # unit seconds
force_measurements = np.array([0.3, 0.4, 0.5, 0.7, 0.8, 0.9])  # unit Newtons

displacement = 100 * 10**-3 # unit m
plate_area = (100 * 10**-3)**2  # unit m^2
flow_height = 1.6 * 10**-3  # unit m

velocity = displacement / time_measurements
dudy = velocity / flow_height
shear_stress = force_measurements / plate_area

A_matrix = dudy[:, np.newaxis]
m, _, _, _ = np.linalg.lstsq(A_matrix, shear_stress, rcond=None)
m = m[0]

print(f"The estimate of the viscosity is mu = {m:.1f} Pa s")
```

```python
plt.scatter(dudy, shear_stress, color="red", label="experimental data")
plt.plot(dudy, dudy*m, label="Linear Fit")
plt.xlabel(r"$\frac{du}{dy} \: [s^{-1}]$")
plt.ylabel(r"$\tau_{yx} \: [N/m^2]$")
plt.text(14, 85, f"y = {m:.3f}x")
plt.legend()
plt.show()
```
